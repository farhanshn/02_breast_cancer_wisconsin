```python
# Import libraries required for data preprocessing

from pathlib import Path
import sys

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
# Define the project root directory

PROJECT_ROOT = Path.cwd().parent

# Add the project root to Python's module search path

if str(PROJECT_ROOT) not in sys.path:
    sys.path.append(str(PROJECT_ROOT))
```

```python
# Import dataset path from the project configuration

from src.confingg import BREAST_DATASET
```

```python
df = pd.read_csv(BREAST_DATASET)
```

```python
df_processed = df.copy()
```

```python
# Convert the diagnosis letters into a binary target

df_processed['target'] = df_processed['diagnosis'].map({'M': 1, 'B': 0})
```

```python
# Check the distribution of the new binary target

df_processed['target'].value_counts().sort_index()
```

```python
# Remove the original diagnosis column and the useless columns

df_processed.drop(columns=['diagnosis', 'id', 'Unnamed: 32'], inplace=True)
```

```python
# Separate input features from the target variable

X = df_processed.drop(columns=['target'])
y = df_processed['target']

X
y
```

```python
print('Features Shape:', X.shape)
print('Target Shape:', y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print('X_train:', X_train.shape)
print('X_test:', X_test.shape)
print()
print('Train class ratio:')
print(y_train.value_counts(normalize=True).sort_index())
print()
print('Test class ratio:')
print(y_test.value_counts(normalize=True).sort_index())
```

```python
# Define numerical features for standardization

numerical_features = X_train.columns.tolist()

numerical_features
```

```python
# Create a StandardScaler instance

scaler = StandardScaler()
```

```python
# Learn scaling parameters from training data only

scaler.fit(X_train[numerical_features])
```

```python
# Standardize numerical features in the training dataset

X_train[numerical_features] = scaler.transform(
    X_train[numerical_features]
)
```

```python
# Apply the same scaling parameters to the testing dataset

X_test[numerical_features] = scaler.transform(
    X_test[numerical_features]
)
```

```python
# Check the mean and standard deviation of scaled numerical features

print("Training mean:")
print(X_train[numerical_features].mean())

print("\nTraining standard deviation:")
print(X_train[numerical_features].std())
```

```python
# Display the final shape of training and testing datasets

print("Final X_train shape:", X_train.shape)
print("Final X_test shape:", X_test.shape)
print("Final y_train shape:", y_train.shape)
print("Final y_test shape:", y_test.shape)
```

```python
# Check for missing values after preprocessing

print("Missing values in X_train:")
print(X_train.isnull().sum().sum())

print("\nMissing values in X_test:")
print(X_test.isnull().sum().sum())
```

```python
# Display the first five rows of the final training dataset

X_train.head()
```

```python
# Create the directory for processed datasets

PROCESSED_DATA_DIR = PROJECT_ROOT / "data" / "processed"

PROCESSED_DATA_DIR.mkdir(
    parents=True,
    exist_ok=True
)
```

```python
# Save processed training and testing datasets

X_train.to_csv(
    PROCESSED_DATA_DIR / "X_train.csv",
    index=False
)

X_test.to_csv(
    PROCESSED_DATA_DIR / "X_test.csv",
    index=False
)

y_train.to_csv(
    PROCESSED_DATA_DIR / "y_train.csv",
    index=False
)

y_test.to_csv(
    PROCESSED_DATA_DIR / "y_test.csv",
    index=False
)
```
